Justificación del método: > Para el dataset de Clasificación (sp_volume), el método Iterativo es el más adecuado. A diferencia de la media, que asume que el volumen de 1970 es igual al de 2024, el método iterativo utiliza la correlación entre empresas del mismo sector para estimar los valores faltantes, preservando mejor la volatilidad real del mercado financiero.

```python
from scipy.stats import chi2_contingency

# 1. Creamos categorías rápidas para la prueba
df_c_iter['AAPL_status'] = df_c_iter['AAPL'].apply(lambda x: 'Alto' if x > df_c_iter['AAPL'].median() else 'Bajo')
df_c_iter['MSFT_status'] = df_c_iter['MSFT'].apply(lambda x: 'Alto' if x > df_c_iter['MSFT'].median() else 'Bajo')

# 2. Tabla de contingencia
tabla_c = pd.crosstab(df_c_iter['AAPL_status'], df_c_iter['MSFT_status'])

# 3. Prueba Chi-cuadrado
chi2, p_chi, dof, expected = chi2_contingency(tabla_c)

print("--- Prueba Chi-cuadrado (AAPL vs MSFT como categorías) ---")
print(f"P-value: {p_chi:.4e}")
if p_chi < 0.05:
    print("Interpretación: Existe una dependencia fuerte entre el comportamiento de ambas empresas.")
```

```python
import scipy.stats as stats

# Comparamos Apple con Amazon
corr_s, p_val_s = stats.spearmanr(df_c_iter['AAPL'], df_c_iter['AMZN'])

print("\n--- Prueba de Spearman (AAPL vs AMZN) ---")
print(f"Coeficiente: {corr_s:.4f}")
print(f"P-value: {p_val_s:.4e}")
```
